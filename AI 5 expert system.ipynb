```python
def laptop_expert(turns_on,internet_works,is_slow):
    if turns_on == "no":
        return "please check the charger and power cable"
    elif internet_works == "no":
        return "please check the network connection"
    elif is_slow == "yes":
        return "please clear the previous tabs and restart again"
    else:
        return "no common problem was detected by our expert system"
```

```python
turn_on = input("Does the laptop has connected ?(yes/no)").lower()
internet_works = input("does the laptop is connected to internet ?(yes/no)").lower()
is_slow = input("is the laptop slow ?(yes/no)").lower()

result=laptop_expert(turn_on, internet_works, is_slow)
print("our expert system advice : ")
print(result)
                
```
